## 2.Import

```python
import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

## 3.load data

```python
df = pd.read_csv('/content/NTI-Final-Project/data/processed/processed_data.csv')
```

```python
x_train = pd.read_csv('/content/NTI-Final-Project/data/processed/X_train.csv')
x_test = pd.read_csv('/content/NTI-Final-Project/data/processed/X_test.csv')
y_train = pd.read_csv('/content/NTI-Final-Project/data/processed/y_train.csv')
y_test = pd.read_csv('/content/NTI-Final-Project/data/processed/y_test.csv')
```

## 4.Identify Numerical and Categorical Features

```python
x_num = x_train.select_dtypes(include='number').columns
x_cat = x_train.select_dtypes(include='object').columns
```

## 5.Feature Scaling

```python
scaler = StandardScaler()
x_train[x_num] =scaler.fit_transform(x_train[x_num])
x_test[x_num] = scaler.transform(x_test[x_num])
```

## 6.Encode Categorical Features

```python
encoder = LabelEncoder()
for col in x_cat :
  x_train[col] = encoder.fit_transform(x_train[col])
  x_test[col] = encoder.transform(x_test[col])
```

## 7.Build Linear Regression Model

```python
lr_model = LinearRegression()
```

## 8.Train Model

```python
lr_model.fit(x_train,y_train)
```

## 9.Make Predication

```python
y_pred = lr_model.predict(x_test)
```

```python
y_pred_train = lr_model.predict(x_train)
```

## 10.Model Evaluation

```python
print(r2_score(y_test,y_pred))
```

```python
print(f'MAE on test data : {mean_absolute_error(y_test,y_pred)}')
print(f'MSE on test data : {mean_squared_error(y_test,y_pred)}')
print(f'RMSE on test data : {np.sqrt(mean_squared_error(y_test,y_pred))}')
```

```python
print(f'MAE on train data : {mean_absolute_error(y_train,y_pred_train)}')
print(f'MSE on train data : {mean_squared_error(y_train,y_pred_train)}')
print(f'RMSE on train data : {np.sqrt(mean_squared_error(y_train,y_pred_train))}')
```

## 11.Model Cofficients

```python
intercept = lr_model.intercept_
weights = lr_model.coef_.flatten()

print(f'theta0 : {intercept}')
print()
print('Weights :')

for feature, w in zip(x_train.columns, weights):
    print(f'  {feature} : {w}')
```
